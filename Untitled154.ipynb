```python
import os
import json
import yaml
import joblib
import warnings

import numpy as np
import pandas as pd
import tensorflow as tf

from datetime import datetime

from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

# ============================================================
# STEP 1: DEFINE PROJECT PATH
# ============================================================

PROJECT_PATH = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Spacecraft Mission Success Analysis and Prediction"
)

DATASET_PATH = os.path.join(
    PROJECT_PATH,
    "RS_Session_266_AU_2850_2.csv"
)

os.makedirs(PROJECT_PATH, exist_ok=True)

print("=" * 65)
print("AI-BASED SPACECRAFT MISSION SUCCESS ANALYSIS")
print("=" * 65)

# ============================================================
# STEP 2: LOAD DATASET
# ============================================================

print("\n[1] Loading Dataset...")

df = pd.read_csv(DATASET_PATH)

print("\nDataset Loaded Successfully!")

print("\nDataset Preview:")
print(df.head())

print("\nDataset Shape:", df.shape)

print("\nDataset Columns:")
print(df.columns.tolist())

# ============================================================
# STEP 3: DATA PREPROCESSING
# ============================================================

print("\n[2] Preprocessing Dataset...")

df.columns = df.columns.str.strip()

# Rename columns

df.rename(
    columns={
        "Year": "Year",
        "Total No. of Spacecraft Missions": "Total_Missions",
        "Successful Spacecraft Missions": "Successful_Missions",
        "Failed Spacecraft Missions": "Failed_Missions"
    },
    inplace=True
)

# Extract year from values such as:
# 2024 (as on 10-12-24)

df["Year"] = (
    df["Year"]
    .astype(str)
    .str.extract(r"(\d{4})")[0]
)

df["Year"] = pd.to_numeric(
    df["Year"],
    errors="coerce"
)

# Convert numerical columns

numeric_columns = [
    "Total_Missions",
    "Successful_Missions",
    "Failed_Missions"
]

for col in numeric_columns:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

# Remove missing values

df.dropna(
    subset=["Year"] + numeric_columns,
    inplace=True
)

# Remove duplicate records

df.drop_duplicates(inplace=True)

# Sort by year

df.sort_values(
    by="Year",
    inplace=True
)

df.reset_index(
    drop=True,
    inplace=True
)

# ============================================================
# STEP 4: FEATURE ENGINEERING
# ============================================================

print("\n[3] Performing Feature Engineering...")

# Calculate success rate

df["Success_Rate"] = np.where(
    df["Total_Missions"] > 0,
    (
        df["Successful_Missions"]
        / df["Total_Missions"]
    ) * 100,
    0
)

# Calculate failure rate

df["Failure_Rate"] = np.where(
    df["Total_Missions"] > 0,
    (
        df["Failed_Missions"]
        / df["Total_Missions"]
    ) * 100,
    0
)

# Mission success ratio

df["Success_Ratio"] = np.where(
    df["Total_Missions"] > 0,
    df["Successful_Missions"]
    / df["Total_Missions"],
    0
)

print("\nProcessed Dataset:")
print(df)

# ============================================================
# STEP 5: SELECT FEATURES AND TARGET
# ============================================================

print("\n[4] Preparing Features and Target...")

# Use Year as input feature.
# Using Successful_Missions or Failed_Missions to predict
# Success_Rate would leak the target into the features.

feature_columns = ["Year"]

target_column = "Success_Rate"

X = df[feature_columns].copy()

y = df[target_column].copy()

print("\nInput Features:")
print(X.head())

print("\nTarget Variable:")
print(y.head())

# ============================================================
# STEP 6: TRAIN RANDOM FOREST MODEL
# ============================================================

print("\n[5] Training Random Forest Model...")

rf_model = Pipeline(
    steps=[
        (
            "scaler",
            StandardScaler()
        ),
        (
            "model",
            RandomForestRegressor(
                n_estimators=100,
                max_depth=3,
                random_state=42
            )
        )
    ]
)

rf_model.fit(X, y)

rf_predictions = rf_model.predict(X)

rf_mae = mean_absolute_error(
    y,
    rf_predictions
)

rf_mse = mean_squared_error(
    y,
    rf_predictions
)

rf_rmse = np.sqrt(rf_mse)

rf_r2 = r2_score(
    y,
    rf_predictions
)

print("\nRandom Forest Training Results:")

print("MAE:", rf_mae)
print("MSE:", rf_mse)
print("RMSE:", rf_rmse)
print("R2 Score:", rf_r2)

# ============================================================
# STEP 7: SAVE RANDOM FOREST MODEL AS PKL
# ============================================================

print("\n[6] Saving PKL Model...")

pkl_path = os.path.join(
    PROJECT_PATH,
    "spacecraft_model.pkl"
)

joblib.dump(
    rf_model,
    pkl_path
)

print("PKL Model Saved:", pkl_path)

# ============================================================
# STEP 8: PREPARE DATA FOR DEEP LEARNING
# ============================================================

print("\n[7] Preparing Deep Learning Model...")

# Separate scaler for the neural network

dl_scaler = StandardScaler()

X_scaled = dl_scaler.fit_transform(X)

X_scaled = X_scaled.astype(
    np.float32
)

y_dl = y.to_numpy().astype(
    np.float32
)

# ============================================================
# STEP 9: BUILD DEEP LEARNING MODEL
# ============================================================

tf.keras.utils.set_random_seed(42)

dl_model = Sequential(
    [
        Input(
            shape=(X_scaled.shape[1],)
        ),

        Dense(
            64,
            activation="relu"
        ),

        Dense(
            32,
            activation="relu"
        ),

        Dense(
            16,
            activation="relu"
        ),

        Dense(
            1,
            activation="linear"
        )
    ]
)

dl_model.compile(
    optimizer=Adam(
        learning_rate=0.001
    ),
    loss="mse",
    metrics=["mae"]
)

print("\nDeep Learning Model Architecture:")

dl_model.summary()

# ============================================================
# STEP 10: TRAIN DEEP LEARNING MODEL
# ============================================================

print("\n[8] Training Deep Learning Model...")

early_stopping = EarlyStopping(
    monitor="loss",
    patience=30,
    restore_best_weights=True
)

history = dl_model.fit(
    X_scaled,
    y_dl,
    epochs=500,
    batch_size=2,
    verbose=0,
    callbacks=[early_stopping]
)

print("\nDeep Learning Training Completed!")

# ============================================================
# STEP 11: EVALUATE DEEP LEARNING MODEL
# ============================================================

dl_predictions = dl_model.predict(
    X_scaled,
    verbose=0
).flatten()

dl_mae = mean_absolute_error(
    y_dl,
    dl_predictions
)

dl_mse = mean_squared_error(
    y_dl,
    dl_predictions
)

dl_rmse = np.sqrt(dl_mse)

dl_r2 = r2_score(
    y_dl,
    dl_predictions
)

print("\nDeep Learning Training Results:")

print("MAE:", dl_mae)
print("MSE:", dl_mse)
print("RMSE:", dl_rmse)
print("R2 Score:", dl_r2)

# ============================================================
# STEP 12: SAVE DEEP LEARNING MODEL AS H5
# ============================================================

print("\n[9] Saving H5 Model...")

h5_path = os.path.join(
    PROJECT_PATH,
    "spacecraft_model.h5"
)

dl_model.save(
    h5_path
)

print("H5 Model Saved:", h5_path)

# Save neural network scaler separately

scaler_path = os.path.join(
    PROJECT_PATH,
    "spacecraft_scaler.pkl"
)

joblib.dump(
    dl_scaler,
    scaler_path
)

print("Deep Learning Scaler Saved:", scaler_path)

# ============================================================
# STEP 13: CREATE YAML CONFIGURATION
# ============================================================

print("\n[10] Creating YAML Configuration...")

config = {

    "project": {
        "name": (
            "AI-Based Spacecraft Mission Success "
            "Analysis and Prediction"
        ),

        "version": "1.0",

        "description": (
            "Exploratory spacecraft mission "
            "success rate analysis using "
            "machine learning"
        )
    },

    "dataset": {
        "file": os.path.basename(DATASET_PATH),

        "records": int(len(df)),

        "features": feature_columns,

        "target": target_column
    },

    "preprocessing": {
        "remove_duplicates": True,

        "handle_missing_values": True,

        "feature_scaling": "StandardScaler"
    },

    "random_forest": {
        "algorithm": "RandomForestRegressor",

        "n_estimators": 100,

        "max_depth": 3,

        "random_state": 42,

        "model_file": "spacecraft_model.pkl"
    },

    "deep_learning": {
        "framework": "TensorFlow",

        "model_type": "Sequential",

        "hidden_layers": [64, 32, 16],

        "activation": "relu",

        "optimizer": "Adam",

        "learning_rate": 0.001,

        "epochs": 500,

        "batch_size": 2,

        "model_file": "spacecraft_model.h5",

        "scaler_file": "spacecraft_scaler.pkl"
    },

    "evaluation": {
        "metrics": [
            "MAE",
            "MSE",
            "RMSE",
            "R2"
        ],

        "evaluation_type": "training_only",

        "validated": False
    }
}

yaml_path = os.path.join(
    PROJECT_PATH,
    "config.yaml"
)

with open(
    yaml_path,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

print("YAML File Saved:", yaml_path)

# ============================================================
# STEP 14: CREATE JSON METADATA
# ============================================================

print("\n[11] Creating JSON Metadata...")

total_missions = int(
    df["Total_Missions"].sum()
)

successful_missions = int(
    df["Successful_Missions"].sum()
)

failed_missions = int(
    df["Failed_Missions"].sum()
)

overall_success_rate = (
    successful_missions / total_missions * 100
    if total_missions > 0
    else 0
)

metadata = {

    "project_name": (
        "AI-Based Spacecraft Mission Success "
        "Analysis and Prediction"
    ),

    "created_at": datetime.now().isoformat(),

    "dataset_information": {
        "filename": os.path.basename(DATASET_PATH),

        "total_records": int(len(df)),

        "start_year": int(df["Year"].min()),

        "end_year": int(df["Year"].max()),

        "total_missions": total_missions,

        "successful_missions": successful_missions,

        "failed_missions": failed_missions,

        "overall_success_rate": overall_success_rate
    },

    "features": feature_columns,

    "target": target_column,

    "models": {

        "random_forest": {
            "model_file": "spacecraft_model.pkl",

            "training_mae": float(rf_mae),

            "training_mse": float(rf_mse),

            "training_rmse": float(rf_rmse),

            "training_r2": float(rf_r2)
        },

        "deep_learning": {
            "model_file": "spacecraft_model.h5",

            "scaler_file": "spacecraft_scaler.pkl",

            "epochs_completed": len(
                history.history["loss"]
            ),

            "training_mae": float(dl_mae),

            "training_mse": float(dl_mse),

            "training_rmse": float(dl_rmse),

            "training_r2": float(dl_r2)
        }
    },

    "limitations": {
        "evaluation": "Training data only",

        "independent_test_set": False,

        "reason": (
            "Only five annual observations are "
            "available. Results cannot establish "
            "reliable predictive accuracy."
        ),

        "2024_data": (
            "Partial year through December 10, 2024"
        )
    }
}

json_path = os.path.join(
    PROJECT_PATH,
    "metadata.json"
)

with open(
    json_path,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4
    )

print("JSON File Saved:", json_path)

# ============================================================
# STEP 15: VERIFY GENERATED FILES
# ============================================================

print("\n[12] Verifying Generated Files...")

generated_files = [
    "spacecraft_model.pkl",
    "spacecraft_model.h5",
    "spacecraft_scaler.pkl",
    "config.yaml",
    "metadata.json"
]

for filename in generated_files:

    filepath = os.path.join(
        PROJECT_PATH,
        filename
    )

    if os.path.exists(filepath):

        print(
            "[SUCCESS]",
            filename
        )

    else:

        print(
            "[FAILED]",
            filename
        )

# ============================================================
# STEP 16: DISPLAY FINAL SUMMARY
# ============================================================

print("\n" + "=" * 65)

print("PROJECT TRAINING COMPLETED SUCCESSFULLY")

print("=" * 65)

print("\nDataset Records:", len(df))

print("Total Missions:", total_missions)

print("Successful Missions:", successful_missions)

print("Failed Missions:", failed_missions)

print(
    "Overall Success Rate:",
    round(overall_success_rate, 2),
    "%"
)

print("\nRandom Forest Training RMSE:", round(rf_rmse, 4))

print("Deep Learning Training RMSE:", round(dl_rmse, 4))

print("\nAll Generated Files Saved In:")

print(PROJECT_PATH)

print("\nIMPORTANT:")
print(
    "Metrics are calculated on training data only. "
    "They do not represent validated prediction accuracy."
)

print("\n" + "=" * 65)
```
