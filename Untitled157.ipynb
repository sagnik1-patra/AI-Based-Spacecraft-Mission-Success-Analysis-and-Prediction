```python
import os
import json
import warnings
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import yaml

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ==============================================================
# STEP 1: PROJECT PATH AND CONFIGURATION
# ==============================================================

PROJECT_PATH = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Spacecraft Mission Success Analysis and Prediction"
)

DATASET_PATH = os.path.join(
    PROJECT_PATH,
    "RS_Session_266_AU_2850_2.csv"
)

os.makedirs(PROJECT_PATH, exist_ok=True)

# PSO Parameters
RANDOM_STATE = 42
NUM_PARTICLES = 60
MAX_ITERATIONS = 250

INERTIA_WEIGHT = 0.7
COGNITIVE_COEFFICIENT = 1.5
SOCIAL_COEFFICIENT = 1.5

POLYNOMIAL_DEGREE = 2
REGULARIZATION = 0.01

LOWER_BOUND = -200.0
UPPER_BOUND = 200.0

FORECAST_START = 2025
FORECAST_END = 2030

rng = np.random.default_rng(RANDOM_STATE)

# Graph settings
sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (12, 7)
plt.rcParams["font.size"] = 11
plt.rcParams["axes.titlesize"] = 15

DPI = 300

print("=" * 70)
print("PSO-BASED SPACECRAFT MISSION SUCCESS ANALYSIS")
print("=" * 70)


# ==============================================================
# STEP 2: GRAPH SAVE AND DISPLAY FUNCTION
# ==============================================================

def save_and_show(filename):

    filepath = os.path.join(
        PROJECT_PATH,
        filename
    )

    plt.tight_layout()

    plt.savefig(
        filepath,
        dpi=DPI,
        bbox_inches="tight"
    )

    print("[SAVED]", filepath)

    plt.show()
    plt.close()


# ==============================================================
# STEP 3: LOAD DATASET
# ==============================================================

print("\n[1] Loading Dataset...")

if not os.path.exists(DATASET_PATH):
    raise FileNotFoundError(
        f"Dataset not found: {DATASET_PATH}"
    )

df = pd.read_csv(DATASET_PATH)

df.columns = df.columns.str.strip()

print("\nOriginal Dataset Columns:")
print(df.columns.tolist())

print("\nDataset Preview:")
print(df.head())


# ==============================================================
# STEP 4: DATA PREPROCESSING
# ==============================================================

print("\n[2] Preprocessing Dataset...")

df.rename(
    columns={
        "Total No. of Spacecraft Missions": "Total_Missions",
        "Successful Spacecraft Missions": "Successful_Missions",
        "Failed Spacecraft Missions": "Failed_Missions"
    },
    inplace=True
)

required_columns = [
    "Year",
    "Total_Missions",
    "Successful_Missions",
    "Failed_Missions"
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing columns: {missing_columns}\n"
        f"Available columns: {df.columns.tolist()}"
    )

# Extract year from text such as:
# 2024 (as on 10-12-24)

df["Year"] = (
    df["Year"]
    .astype(str)
    .str.extract(r"(\d{4})")[0]
)

for col in required_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

df.dropna(
    subset=required_columns,
    inplace=True
)

df.drop_duplicates(inplace=True)

df.sort_values(
    by="Year",
    inplace=True
)

df.reset_index(
    drop=True,
    inplace=True
)

df["Year"] = df["Year"].astype(int)

if df["Year"].duplicated().any():
    raise ValueError(
        "Duplicate years detected. "
        "Aggregate annual mission records first."
    )

if len(df) < 3:
    raise ValueError(
        "At least three annual records are required."
    )

# Calculate success and failure rates

df["Success_Rate"] = np.where(
    df["Total_Missions"] > 0,
    (
        df["Successful_Missions"]
        / df["Total_Missions"]
    ) * 100,
    0
)

df["Failure_Rate"] = np.where(
    df["Total_Missions"] > 0,
    (
        df["Failed_Missions"]
        / df["Total_Missions"]
    ) * 100,
    0
)

print("\nProcessed Dataset:")
print(df.to_string(index=False))


# ==============================================================
# STEP 5: FEATURE ENGINEERING
# ==============================================================

print("\n[3] Preparing Features...")

# Center and scale years for stable optimization

YEAR_CENTER = float(
    df["Year"].mean()
)

YEAR_SCALE = 10.0

def create_features(years):

    years = np.asarray(
        years,
        dtype=float
    )

    normalized_years = (
        years - YEAR_CENTER
    ) / YEAR_SCALE

    features = [
        np.ones_like(normalized_years)
    ]

    for degree in range(
        1,
        POLYNOMIAL_DEGREE + 1
    ):
        features.append(
            normalized_years ** degree
        )

    return np.column_stack(features)


X = create_features(
    df["Year"].to_numpy()
)

y = df[
    "Success_Rate"
].to_numpy(dtype=float)

NUM_DIMENSIONS = X.shape[1]

print("Feature Matrix Shape:", X.shape)
print("Target Shape:", y.shape)
print("Number of PSO Dimensions:", NUM_DIMENSIONS)


# ==============================================================
# STEP 6: DEFINE PREDICTION FUNCTION
# ==============================================================

def predict_raw(features, coefficients):

    return features @ coefficients


def predict_success_rate(features, coefficients):

    predictions = predict_raw(
        features,
        coefficients
    )

    # Success rate must remain between 0 and 100
    return np.clip(
        predictions,
        0,
        100
    )


# ==============================================================
# STEP 7: DEFINE PSO FITNESS FUNCTION
# ==============================================================

print("\n[4] Defining PSO Fitness Function...")

def fitness_function(coefficients):

    predictions = predict_raw(
        X,
        coefficients
    )

    mse = np.mean(
        (y - predictions) ** 2
    )

    # Regularize non-intercept coefficients
    penalty = REGULARIZATION * np.sum(
        coefficients[1:] ** 2
    )

    return float(
        mse + penalty
    )


# ==============================================================
# STEP 8: INITIALIZE PSO PARTICLES
# ==============================================================

print("\n[5] Initializing PSO Particles...")

positions = rng.uniform(
    LOWER_BOUND,
    UPPER_BOUND,
    size=(NUM_PARTICLES, NUM_DIMENSIONS)
)

velocities = rng.uniform(
    -1,
    1,
    size=(NUM_PARTICLES, NUM_DIMENSIONS)
)

# Initialize one particle with mean baseline

positions[0] = np.zeros(
    NUM_DIMENSIONS
)

positions[0, 0] = float(
    np.mean(y)
)

# Initialize one particle with least-squares solution

least_squares_solution = np.linalg.lstsq(
    X,
    y,
    rcond=None
)[0]

positions[1] = np.clip(
    least_squares_solution,
    LOWER_BOUND,
    UPPER_BOUND
)

# Personal best positions

personal_best_positions = positions.copy()

personal_best_scores = np.array([
    fitness_function(position)
    for position in positions
])

# Global best particle

best_index = int(
    np.argmin(personal_best_scores)
)

global_best_position = (
    personal_best_positions[best_index].copy()
)

global_best_score = float(
    personal_best_scores[best_index]
)

# Store optimization history

fitness_history = []
mean_fitness_history = []


# ==============================================================
# STEP 9: PARTICLE SWARM OPTIMIZATION
# ==============================================================

print("\n[6] Running Particle Swarm Optimization...")

for iteration in range(MAX_ITERATIONS):

    for i in range(NUM_PARTICLES):

        # Evaluate current particle
        current_fitness = fitness_function(
            positions[i]
        )

        # Update personal best
        if current_fitness < personal_best_scores[i]:

            personal_best_scores[i] = current_fitness

            personal_best_positions[i] = (
                positions[i].copy()
            )

        # Update global best
        if current_fitness < global_best_score:

            global_best_score = float(
                current_fitness
            )

            global_best_position = (
                positions[i].copy()
            )

    # Save convergence history
    fitness_history.append(
        global_best_score
    )

    mean_fitness_history.append(
        float(
            np.mean([
                fitness_function(position)
                for position in positions
            ])
        )
    )

    # Update particle velocities and positions
    for i in range(NUM_PARTICLES):

        r1 = rng.random(
            NUM_DIMENSIONS
        )

        r2 = rng.random(
            NUM_DIMENSIONS
        )

        cognitive_velocity = (
            COGNITIVE_COEFFICIENT
            * r1
            * (
                personal_best_positions[i]
                - positions[i]
            )
        )

        social_velocity = (
            SOCIAL_COEFFICIENT
            * r2
            * (
                global_best_position
                - positions[i]
            )
        )

        velocities[i] = (
            INERTIA_WEIGHT * velocities[i]
            + cognitive_velocity
            + social_velocity
        )

        # Limit velocity to prevent instability
        velocities[i] = np.clip(
            velocities[i],
            -20,
            20
        )

        # Update position
        positions[i] += velocities[i]

        # Apply coefficient bounds
        positions[i] = np.clip(
            positions[i],
            LOWER_BOUND,
            UPPER_BOUND
        )

    if (iteration + 1) % 25 == 0:

        print(
            f"Iteration {iteration + 1:3d} | "
            f"Best Fitness: {global_best_score:.6f}"
        )

# Final fitness evaluation

for i in range(NUM_PARTICLES):

    current_fitness = fitness_function(
        positions[i]
    )

    if current_fitness < global_best_score:

        global_best_score = float(
            current_fitness
        )

        global_best_position = (
            positions[i].copy()
        )

print("\nPSO Optimization Completed!")

print(
    "Best Fitness:",
    global_best_score
)

print(
    "Best Coefficients:",
    global_best_position
)


# ==============================================================
# STEP 10: GENERATE HISTORICAL PREDICTIONS
# ==============================================================

print("\n[7] Generating PSO Historical Predictions...")

pso_predictions = predict_success_rate(
    X,
    global_best_position
)

absolute_errors = np.abs(
    y - pso_predictions
)

squared_errors = (
    y - pso_predictions
) ** 2

# ==============================================================
# STEP 11: CALCULATE TRAINING METRICS
# ==============================================================

print("\n[8] Calculating PSO Training Metrics...")

pso_mae = mean_absolute_error(
    y,
    pso_predictions
)

pso_mse = mean_squared_error(
    y,
    pso_predictions
)

pso_rmse = np.sqrt(
    pso_mse
)

pso_r2 = r2_score(
    y,
    pso_predictions
)

print("\nPSO Training Metrics:")
print("MAE:", pso_mae)
print("MSE:", pso_mse)
print("RMSE:", pso_rmse)
print("R2 Score:", pso_r2)


# ==============================================================
# STEP 12: SAVE PSO MODEL
# ==============================================================

print("\n[9] Saving PSO Model...")

model_data = {

    "algorithm": "Particle Swarm Optimization",

    "model_type": "Polynomial Regression",

    "coefficients": global_best_position,

    "polynomial_degree": POLYNOMIAL_DEGREE,

    "year_center": YEAR_CENTER,

    "year_scale": YEAR_SCALE,

    "best_fitness": global_best_score,

    "training_years": df["Year"].tolist(),

    "training_only": True
}

model_path = os.path.join(
    PROJECT_PATH,
    "pso_model.pkl"
)

joblib.dump(
    model_data,
    model_path
)

print("[SAVED]", model_path)


# ==============================================================
# STEP 13: PSO ACCURACY GRAPH
# ==============================================================

print("\n[10] Generating PSO Accuracy Graph...")

fig, axes = plt.subplots(
    1,
    2,
    figsize=(15, 6)
)

# Graph 1: Optimization convergence

axes[0].plot(
    range(1, MAX_ITERATIONS + 1),
    fitness_history,
    linewidth=2,
    color="#2980b9",
    label="Global Best Fitness"
)

axes[0].plot(
    range(1, MAX_ITERATIONS + 1),
    mean_fitness_history,
    linewidth=1.5,
    color="#e67e22",
    alpha=0.7,
    label="Mean Swarm Fitness"
)

axes[0].set_title(
    "PSO Optimization Convergence"
)

axes[0].set_xlabel(
    "Iteration"
)

axes[0].set_ylabel(
    "Fitness (Regularized MSE)"
)

axes[0].legend()
axes[0].grid(alpha=0.3)

# Graph 2: Training errors

metric_names = [
    "MAE",
    "RMSE"
]

metric_values = [
    pso_mae,
    pso_rmse
]

bars = axes[1].bar(
    metric_names,
    metric_values,
    color=["#27ae60", "#e74c3c"],
    width=0.5
)

axes[1].set_title(
    "PSO Training Error Metrics"
)

axes[1].set_ylabel(
    "Error (Percentage Points)"
)

axes[1].set_ylim(
    0,
    max(metric_values + [1]) * 1.25
)

for bar in bars:

    height = bar.get_height()

    axes[1].text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"{height:.2f}",
        ha="center",
        va="bottom"
    )

fig.suptitle(
    "PSO Model Training Performance",
    fontsize=16
)

save_and_show(
    "pso_accuracy_graph.png"
)


# ==============================================================
# STEP 14: PSO CORRELATION HEATMAP
# ==============================================================

print("\n[11] Generating PSO Heatmap...")

heatmap_columns = [
    "Year",
    "Total_Missions",
    "Successful_Missions",
    "Failed_Missions",
    "Success_Rate",
    "Failure_Rate"
]

correlation_matrix = df[
    heatmap_columns
].corr()

plt.figure(
    figsize=(11, 9)
)

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
    square=True,
    vmin=-1,
    vmax=1
)

plt.title(
    "PSO Spacecraft Mission "
    "Correlation Heatmap"
)

save_and_show(
    "pso_heatmap.png"
)


# ==============================================================
# STEP 15: PSO COMPARISON GRAPH
# ==============================================================

print("\n[12] Generating PSO Comparison Graph...")

years = df[
    "Year"
].to_numpy()

plt.figure(
    figsize=(13, 7)
)

plt.plot(
    years,
    y,
    marker="o",
    linewidth=3,
    color="black",
    label="Actual Success Rate"
)

plt.plot(
    years,
    pso_predictions,
    marker="s",
    linestyle="--",
    linewidth=2,
    color="#8e44ad",
    label="PSO Fitted Success Rate"
)

plt.xlabel("Year")
plt.ylabel("Mission Success Rate (%)")

plt.title(
    "Actual vs PSO-Fitted "
    "Spacecraft Mission Success Rate"
)

plt.ylim(0, 110)
plt.xticks(years)
plt.legend()
plt.grid(alpha=0.3)

save_and_show(
    "pso_comparison_graph.png"
)


# ==============================================================
# STEP 16: GENERATE PSO RESULT CSV
# ==============================================================

print("\n[13] Generating PSO Result CSV...")

result_df = pd.DataFrame({

    "Year": df["Year"],

    "Total_Missions":
        df["Total_Missions"],

    "Successful_Missions":
        df["Successful_Missions"],

    "Failed_Missions":
        df["Failed_Missions"],

    "Actual_Success_Rate":
        df["Success_Rate"],

    "Actual_Failure_Rate":
        df["Failure_Rate"],

    "PSO_Predicted_Success_Rate":
        pso_predictions,

    "PSO_Absolute_Error":
        absolute_errors,

    "PSO_Squared_Error":
        squared_errors
})

result_df = result_df.round(4)

result_path = os.path.join(
    PROJECT_PATH,
    "pso_result.csv"
)

result_df.to_csv(
    result_path,
    index=False
)

print("[SAVED]", result_path)

print("\nHistorical Results:")
print(result_df.to_string(index=False))


# ==============================================================
# STEP 17: PSO RESULT GRAPH
# ==============================================================

print("\n[14] Generating PSO Result Graph...")

successful = df[
    "Successful_Missions"
].to_numpy()

failed = df[
    "Failed_Missions"
].to_numpy()

plt.figure(
    figsize=(13, 7)
)

plt.bar(
    years,
    successful,
    color="#27ae60",
    label="Successful Missions"
)

plt.bar(
    years,
    failed,
    bottom=successful,
    color="#e74c3c",
    label="Failed Missions"
)

for i in range(len(years)):

    total = successful[i] + failed[i]

    plt.text(
        years[i],
        total + 0.1,
        str(int(total)),
        ha="center",
        fontweight="bold"
    )

plt.xlabel("Year")
plt.ylabel("Number of Missions")

plt.title(
    "PSO Project - Historical "
    "Spacecraft Mission Results"
)

plt.xticks(years)
plt.legend()
plt.grid(axis="y", alpha=0.3)

save_and_show(
    "pso_result_graph.png"
)


# ==============================================================
# STEP 18: FUTURE PREDICTION USING PSO
# ==============================================================

print("\n[15] Generating PSO Future Predictions...")

future_years = np.arange(
    FORECAST_START,
    FORECAST_END + 1
)

future_X = create_features(
    future_years
)

future_predictions = predict_success_rate(
    future_X,
    global_best_position
)

print("\nExperimental Future Predictions:")

for year, prediction in zip(
    future_years,
    future_predictions
):

    print(
        f"{year}: {prediction:.2f}%"
    )


# ==============================================================
# STEP 19: SAVE PSO PREDICTION CSV
# ==============================================================

print("\n[16] Generating PSO Prediction CSV...")

prediction_df = pd.DataFrame({

    "Year": future_years,

    "PSO_Predicted_Success_Rate":
        np.round(future_predictions, 4),

    "Prediction_Type":
        ["Experimental Forecast"] * len(future_years),

    "Validated":
        [False] * len(future_years)
})

prediction_path = os.path.join(
    PROJECT_PATH,
    "pso_prediction.csv"
)

prediction_df.to_csv(
    prediction_path,
    index=False
)

print("[SAVED]", prediction_path)

print("\nPrediction Results:")
print(prediction_df.to_string(index=False))


# ==============================================================
# STEP 20: PSO PREDICTION GRAPH
# ==============================================================

print("\n[17] Generating PSO Prediction Graph...")

plt.figure(
    figsize=(14, 8)
)

# Historical success rates
plt.plot(
    years,
    y,
    marker="o",
    linewidth=3,
    color="black",
    label="Historical Success Rate"
)

# Future PSO predictions
plt.plot(
    future_years,
    future_predictions,
    marker="s",
    linestyle="--",
    linewidth=2,
    color="#8e44ad",
    label="PSO Experimental Forecast"
)

# Forecast boundary
plt.axvline(
    x=2024.5,
    linestyle="--",
    color="gray",
    linewidth=2,
    label="Forecast Boundary"
)

plt.axvspan(
    2024.5,
    FORECAST_END + 0.5,
    color="orange",
    alpha=0.08
)

plt.xlabel("Year")
plt.ylabel("Mission Success Rate (%)")

plt.title(
    "PSO-Based Spacecraft Mission "
    "Success Rate Forecast (2025-2030)\n"
    "Experimental Projections - Not Validated"
)

plt.xticks(
    np.arange(
        int(years.min()),
        FORECAST_END + 1
    )
)

plt.ylim(0, 110)
plt.legend()
plt.grid(alpha=0.3)

save_and_show(
    "pso_prediction_graph.png"
)


# ==============================================================
# STEP 21: SAVE PSO MODEL METRICS
# ==============================================================

print("\n[18] Saving PSO Model Metrics...")

metrics_df = pd.DataFrame({

    "Model": ["PSO"],

    "Training_MAE": [pso_mae],

    "Training_MSE": [pso_mse],

    "Training_RMSE": [pso_rmse],

    "Training_R2": [pso_r2],

    "Best_Fitness": [global_best_score],

    "Iterations": [MAX_ITERATIONS],

    "Particles": [NUM_PARTICLES],

    "Evaluation_Type": ["Training Only"]
})

metrics_path = os.path.join(
    PROJECT_PATH,
    "pso_model_comparison_metrics.csv"
)

metrics_df.to_csv(
    metrics_path,
    index=False
)

print("[SAVED]", metrics_path)


# ==============================================================
# STEP 22: SAVE PSO YAML CONFIGURATION
# ==============================================================

print("\n[19] Saving PSO Configuration...")

config = {

    "project": {
        "name": (
            "AI-Based Spacecraft Mission "
            "Success Analysis and Prediction"
        ),
        "algorithm": "Particle Swarm Optimization",
        "version": "1.0"
    },

    "dataset": {
        "filename": os.path.basename(DATASET_PATH),
        "records": int(len(df)),
        "features": ["Year"],
        "target": "Success_Rate"
    },

    "pso_parameters": {
        "particles": NUM_PARTICLES,
        "iterations": MAX_ITERATIONS,
        "inertia_weight": INERTIA_WEIGHT,
        "cognitive_coefficient": COGNITIVE_COEFFICIENT,
        "social_coefficient": SOCIAL_COEFFICIENT,
        "polynomial_degree": POLYNOMIAL_DEGREE,
        "regularization": REGULARIZATION,
        "lower_bound": LOWER_BOUND,
        "upper_bound": UPPER_BOUND,
        "random_state": RANDOM_STATE
    },

    "forecast": {
        "start_year": FORECAST_START,
        "end_year": FORECAST_END,
        "validated": False
    },

    "model": {
        "filename": "pso_model.pkl",
        "type": "PSO-Optimized Polynomial Regression"
    },

    "evaluation": {
        "type": "Training Only",
        "independent_validation": False
    }
}

yaml_path = os.path.join(
    PROJECT_PATH,
    "pso_config.yaml"
)

with open(
    yaml_path,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

print("[SAVED]", yaml_path)


# ==============================================================
# STEP 23: SAVE PSO JSON METADATA
# ==============================================================

print("\n[20] Saving PSO Metadata...")

total_missions = int(
    df["Total_Missions"].sum()
)

successful_missions = int(
    df["Successful_Missions"].sum()
)

failed_missions = int(
    df["Failed_Missions"].sum()
)

overall_success_rate = (
    successful_missions / total_missions * 100
    if total_missions > 0
    else 0.0
)

metadata = {

    "project_name": (
        "AI-Based Spacecraft Mission "
        "Success Analysis and Prediction"
    ),

    "algorithm": "Particle Swarm Optimization",

    "created_at": datetime.now().isoformat(),

    "dataset": {
        "records": int(len(df)),
        "start_year": int(df["Year"].min()),
        "end_year": int(df["Year"].max()),
        "total_missions": total_missions,
        "successful_missions": successful_missions,
        "failed_missions": failed_missions,
        "overall_success_rate": float(overall_success_rate)
    },

    "model": {
        "type": "Polynomial Regression",
        "optimization": "PSO",
        "coefficients": global_best_position.tolist(),
        "degree": POLYNOMIAL_DEGREE,
        "best_fitness": float(global_best_score)
    },

    "training_metrics": {
        "MAE": float(pso_mae),
        "MSE": float(pso_mse),
        "RMSE": float(pso_rmse),
        "R2": float(pso_r2)
    },

    "limitations": {
        "training_only": True,
        "validated_forecasts": False,
        "small_dataset": True,
        "partial_2024_observation": True,
        "note": (
            "Only five annual observations are available. "
            "Training metrics are not independent validation. "
            "Forecasts are exploratory."
        )
    },

    "generated_files": [
        "pso_model.pkl",
        "pso_config.yaml",
        "pso_metadata.json",
        "pso_accuracy_graph.png",
        "pso_heatmap.png",
        "pso_comparison_graph.png",
        "pso_result.csv",
        "pso_result_graph.png",
        "pso_prediction.csv",
        "pso_prediction_graph.png",
        "pso_model_comparison_metrics.csv"
    ]
}

metadata_path = os.path.join(
    PROJECT_PATH,
    "pso_metadata.json"
)

with open(
    metadata_path,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4
    )

print("[SAVED]", metadata_path)


# ==============================================================
# STEP 24: FINAL VERIFICATION
# ==============================================================

print("\n" + "=" * 70)
print("PSO SPACECRAFT ANALYSIS COMPLETED")
print("=" * 70)

generated_files = [
    "pso_model.pkl",
    "pso_config.yaml",
    "pso_metadata.json",
    "pso_accuracy_graph.png",
    "pso_heatmap.png",
    "pso_comparison_graph.png",
    "pso_result.csv",
    "pso_result_graph.png",
    "pso_prediction.csv",
    "pso_prediction_graph.png",
    "pso_model_comparison_metrics.csv"
]

for filename in generated_files:

    filepath = os.path.join(
        PROJECT_PATH,
        filename
    )

    if os.path.exists(filepath):
        print("[SUCCESS]", filename)
    else:
        print("[FAILED]", filename)

print("\nDATASET SUMMARY")
print("Total Records:", len(df))
print("Total Missions:", total_missions)
print("Successful Missions:", successful_missions)
print("Failed Missions:", failed_missions)
print(
    "Overall Success Rate:",
    round(overall_success_rate, 2),
    "%"
)

print("\nPSO TRAINING PERFORMANCE")
print("MAE:", round(pso_mae, 4))
print("MSE:", round(pso_mse, 4))
print("RMSE:", round(pso_rmse, 4))
print("R2:", round(pso_r2, 4))
print("Best Fitness:", round(global_best_score, 4))

print("\nAll files saved in:")
print(PROJECT_PATH)

print("\nIMPORTANT:")
print("Metrics represent training fit only.")
print("Future forecasts are experimental and not validated.")

print("\n" + "=" * 70)
print("ALL PSO RESULTS GENERATED SUCCESSFULLY")
print("=" * 70)
```
