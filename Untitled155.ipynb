```python
import os
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import tensorflow as tf

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ============================================================
# STEP 1: PROJECT PATH
# ============================================================

PROJECT_PATH = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Spacecraft Mission Success Analysis and Prediction"
)

DATASET_PATH = os.path.join(
    PROJECT_PATH,
    "RS_Session_266_AU_2850_2.csv"
)

PKL_PATH = os.path.join(
    PROJECT_PATH,
    "spacecraft_model.pkl"
)

H5_PATH = os.path.join(
    PROJECT_PATH,
    "spacecraft_model.h5"
)

SCALER_PATH = os.path.join(
    PROJECT_PATH,
    "spacecraft_scaler.pkl"
)

os.makedirs(PROJECT_PATH, exist_ok=True)

# ============================================================
# STEP 2: GRAPH SETTINGS
# ============================================================

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (12, 7)
plt.rcParams["font.size"] = 11
plt.rcParams["axes.titlesize"] = 15
plt.rcParams["axes.labelsize"] = 12

DPI = 300

def save_and_show(filename):

    filepath = os.path.join(
        PROJECT_PATH,
        filename
    )

    plt.tight_layout()

    plt.savefig(
        filepath,
        dpi=DPI,
        bbox_inches="tight"
    )

    print("\nSaved:", filepath)

    plt.show()

    plt.close()


# ============================================================
# STEP 3: LOAD DATASET
# ============================================================

print("=" * 70)
print("AI-BASED SPACECRAFT MISSION SUCCESS ANALYSIS")
print("=" * 70)

print("\n[1] Loading Dataset...")

df = pd.read_csv(DATASET_PATH)

df.columns = df.columns.str.strip()

print("\nOriginal Columns:")
print(df.columns.tolist())

# ============================================================
# STEP 4: DATA PREPROCESSING
# ============================================================

print("\n[2] Preprocessing Dataset...")

df.rename(
    columns={
        "Total No. of Spacecraft Missions":
            "Total_Missions",

        "Successful Spacecraft Missions":
            "Successful_Missions",

        "Failed Spacecraft Missions":
            "Failed_Missions"
    },
    inplace=True
)

required_columns = [
    "Year",
    "Total_Missions",
    "Successful_Missions",
    "Failed_Missions"
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:

    raise ValueError(
        "Missing columns: "
        + str(missing_columns)
    )

# Extract year

df["Year"] = (
    df["Year"]
    .astype(str)
    .str.extract(r"(\d{4})")[0]
)

df["Year"] = pd.to_numeric(
    df["Year"],
    errors="coerce"
)

# Convert numerical columns

numeric_columns = [
    "Total_Missions",
    "Successful_Missions",
    "Failed_Missions"
]

for col in numeric_columns:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

df.dropna(
    subset=required_columns,
    inplace=True
)

df.drop_duplicates(inplace=True)

df.sort_values(
    by="Year",
    inplace=True
)

df.reset_index(
    drop=True,
    inplace=True
)

df["Year"] = df["Year"].astype(int)

# Calculate success rate

df["Success_Rate"] = np.where(
    df["Total_Missions"] > 0,
    (
        df["Successful_Missions"]
        / df["Total_Missions"]
    ) * 100,
    0
)

# Calculate failure rate

df["Failure_Rate"] = np.where(
    df["Total_Missions"] > 0,
    (
        df["Failed_Missions"]
        / df["Total_Missions"]
    ) * 100,
    0
)

print("\nProcessed Dataset:")
print(df)

# ============================================================
# STEP 5: LOAD TRAINED MODELS
# ============================================================

print("\n[3] Loading Trained Models...")

rf_model = joblib.load(
    PKL_PATH
)

dl_model = tf.keras.models.load_model(
    H5_PATH,
    compile=False
)

dl_scaler = joblib.load(
    SCALER_PATH
)

print("\nRandom Forest Model Loaded!")
print("Deep Learning Model Loaded!")
print("Scaler Loaded!")

# ============================================================
# STEP 6: PREPARE INPUT FEATURES
# ============================================================

print("\n[4] Preparing Input Features...")

feature_columns = ["Year"]

X = df[feature_columns].copy()

y = df["Success_Rate"].to_numpy()

X_scaled = dl_scaler.transform(X)

X_scaled = X_scaled.astype(
    np.float32
)

# ============================================================
# STEP 7: GENERATE MODEL PREDICTIONS
# ============================================================

print("\n[5] Generating Model Predictions...")

rf_predictions = rf_model.predict(X)

dl_predictions = dl_model.predict(
    X_scaled,
    verbose=0
).flatten()

# Success rate is bounded between 0 and 100

rf_predictions = np.clip(
    rf_predictions,
    0,
    100
)

dl_predictions = np.clip(
    dl_predictions,
    0,
    100
)

print("\nRandom Forest Predictions:")
print(rf_predictions)

print("\nDeep Learning Predictions:")
print(dl_predictions)

# ============================================================
# STEP 8: CALCULATE TRAINING METRICS
# ============================================================

print("\n[6] Calculating Training Metrics...")

rf_mae = mean_absolute_error(
    y,
    rf_predictions
)

rf_mse = mean_squared_error(
    y,
    rf_predictions
)

rf_rmse = np.sqrt(rf_mse)

rf_r2 = r2_score(
    y,
    rf_predictions
)

dl_mae = mean_absolute_error(
    y,
    dl_predictions
)

dl_mse = mean_squared_error(
    y,
    dl_predictions
)

dl_rmse = np.sqrt(dl_mse)

dl_r2 = r2_score(
    y,
    dl_predictions
)

print("\nRandom Forest Training Metrics:")
print("MAE:", rf_mae)
print("MSE:", rf_mse)
print("RMSE:", rf_rmse)
print("R2:", rf_r2)

print("\nDeep Learning Training Metrics:")
print("MAE:", dl_mae)
print("MSE:", dl_mse)
print("RMSE:", dl_rmse)
print("R2:", dl_r2)

# ============================================================
# STEP 9: ACCURACY GRAPH
# ============================================================

print("\n[7] Generating Accuracy Graph...")

# For regression, report R2 and error metrics.
# R2 is not classification accuracy.

models = [
    "Random Forest",
    "Deep Learning"
]

mae_values = [
    rf_mae,
    dl_mae
]

rmse_values = [
    rf_rmse,
    dl_rmse
]

x = np.arange(len(models))

width = 0.35

plt.figure(figsize=(11, 7))

bars1 = plt.bar(
    x - width / 2,
    mae_values,
    width,
    label="Training MAE",
    color="#3498db"
)

bars2 = plt.bar(
    x + width / 2,
    rmse_values,
    width,
    label="Training RMSE",
    color="#e74c3c"
)

plt.xlabel("Machine Learning Models")

plt.ylabel("Error (Percentage Points)")

plt.title(
    "Spacecraft Mission Prediction - "
    "Training Performance"
)

plt.xticks(
    x,
    models
)

plt.legend()

plt.grid(
    axis="y",
    alpha=0.3
)

for bars in [bars1, bars2]:

    for bar in bars:

        height = bar.get_height()

        plt.text(
            bar.get_x() + bar.get_width() / 2,
            height,
            f"{height:.2f}",
            ha="center",
            va="bottom"
        )

save_and_show(
    "accuracy_graph.png"
)

# ============================================================
# STEP 10: CORRELATION HEATMAP
# ============================================================

print("\n[8] Generating Correlation Heatmap...")

heatmap_columns = [
    "Year",
    "Total_Missions",
    "Successful_Missions",
    "Failed_Missions",
    "Success_Rate",
    "Failure_Rate"
]

correlation_matrix = df[
    heatmap_columns
].corr()

plt.figure(figsize=(12, 9))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
    square=True,
    vmin=-1,
    vmax=1
)

plt.title(
    "Spacecraft Mission Dataset "
    "Correlation Heatmap"
)

save_and_show(
    "heatmap.png"
)

# ============================================================
# STEP 11: COMPARISON GRAPH
# ============================================================

print("\n[9] Generating Comparison Graph...")

plt.figure(figsize=(13, 7))

plt.plot(
    df["Year"],
    y,
    marker="o",
    linewidth=3,
    color="black",
    label="Actual Success Rate"
)

plt.plot(
    df["Year"],
    rf_predictions,
    marker="s",
    linestyle="--",
    linewidth=2,
    color="#2980b9",
    label="Random Forest Prediction"
)

plt.plot(
    df["Year"],
    dl_predictions,
    marker="^",
    linestyle="--",
    linewidth=2,
    color="#e74c3c",
    label="Deep Learning Prediction"
)

plt.xlabel("Year")

plt.ylabel("Mission Success Rate (%)")

plt.title(
    "Actual vs Predicted Spacecraft "
    "Mission Success Rate"
)

plt.ylim(0, 110)

plt.xticks(
    df["Year"]
)

plt.legend()

plt.grid(
    alpha=0.3
)

save_and_show(
    "comparison_graph.png"
)

# ============================================================
# STEP 12: GENERATE RESULT CSV
# ============================================================

print("\n[10] Generating Result CSV...")

result_df = pd.DataFrame({

    "Year": df["Year"],

    "Total_Missions":
        df["Total_Missions"],

    "Successful_Missions":
        df["Successful_Missions"],

    "Failed_Missions":
        df["Failed_Missions"],

    "Actual_Success_Rate":
        df["Success_Rate"],

    "Actual_Failure_Rate":
        df["Failure_Rate"],

    "Random_Forest_Prediction":
        rf_predictions,

    "Deep_Learning_Prediction":
        dl_predictions,

    "RF_Absolute_Error":
        np.abs(y - rf_predictions),

    "DL_Absolute_Error":
        np.abs(y - dl_predictions)

})

result_path = os.path.join(
    PROJECT_PATH,
    "result.csv"
)

result_df.to_csv(
    result_path,
    index=False
)

print("\nResult CSV Saved:")
print(result_path)

print("\nResult Preview:")
print(result_df)

# ============================================================
# STEP 13: RESULT GRAPH
# ============================================================

print("\n[11] Generating Result Graph...")

years = df["Year"].to_numpy()

successful = df[
    "Successful_Missions"
].to_numpy()

failed = df[
    "Failed_Missions"
].to_numpy()

plt.figure(figsize=(13, 7))

plt.bar(
    years,
    successful,
    color="#27ae60",
    label="Successful Missions"
)

plt.bar(
    years,
    failed,
    bottom=successful,
    color="#e74c3c",
    label="Failed Missions"
)

for i in range(len(years)):

    plt.text(
        years[i],
        successful[i] + failed[i] + 0.1,
        str(int(successful[i] + failed[i])),
        ha="center",
        fontweight="bold"
    )

plt.xlabel("Year")

plt.ylabel("Number of Missions")

plt.title(
    "Year-Wise Spacecraft Mission "
    "Success and Failure Analysis"
)

plt.xticks(years)

plt.legend()

plt.grid(
    axis="y",
    alpha=0.3
)

save_and_show(
    "result_graph.png"
)

# ============================================================
# STEP 14: FUTURE PREDICTIONS
# ============================================================

print("\n[12] Generating Future Predictions...")

future_years = np.arange(
    2025,
    2031
)

future_df = pd.DataFrame({
    "Year": future_years
})

future_X = future_df[
    feature_columns
]

# Random Forest forecast

future_rf = rf_model.predict(
    future_X
)

# Deep Learning forecast

future_scaled = dl_scaler.transform(
    future_X
)

future_scaled = future_scaled.astype(
    np.float32
)

future_dl = dl_model.predict(
    future_scaled,
    verbose=0
).flatten()

# Bound success rates between 0 and 100

future_rf = np.clip(
    future_rf,
    0,
    100
)

future_dl = np.clip(
    future_dl,
    0,
    100
)

# Average prediction for comparison only

future_ensemble = (
    future_rf + future_dl
) / 2

# ============================================================
# STEP 15: GENERATE PREDICTION CSV
# ============================================================

print("\n[13] Generating Prediction CSV...")

prediction_df = pd.DataFrame({

    "Year": future_years,

    "Random_Forest_Prediction":
        future_rf,

    "Deep_Learning_Prediction":
        future_dl,

    "Ensemble_Prediction":
        future_ensemble

})

prediction_df[
    "Random_Forest_Prediction"
] = prediction_df[
    "Random_Forest_Prediction"
].round(2)

prediction_df[
    "Deep_Learning_Prediction"
] = prediction_df[
    "Deep_Learning_Prediction"
].round(2)

prediction_df[
    "Ensemble_Prediction"
] = prediction_df[
    "Ensemble_Prediction"
].round(2)

prediction_df[
    "Prediction_Type"
] = "Experimental Forecast"

prediction_path = os.path.join(
    PROJECT_PATH,
    "prediction.csv"
)

prediction_df.to_csv(
    prediction_path,
    index=False
)

print("\nPrediction CSV Saved:")
print(prediction_path)

print("\nFuture Prediction Results:")
print(prediction_df)

# ============================================================
# STEP 16: PREDICTION GRAPH
# ============================================================

print("\n[14] Generating Prediction Graph...")

plt.figure(figsize=(14, 8))

# Historical data

plt.plot(
    df["Year"],
    df["Success_Rate"],
    marker="o",
    linewidth=3,
    color="black",
    label="Historical Success Rate"
)

# Random Forest forecast

plt.plot(
    future_years,
    future_rf,
    marker="s",
    linestyle="--",
    linewidth=2,
    color="#3498db",
    label="Random Forest Forecast"
)

# Deep Learning forecast

plt.plot(
    future_years,
    future_dl,
    marker="^",
    linestyle="--",
    linewidth=2,
    color="#e74c3c",
    label="Deep Learning Forecast"
)

# Ensemble forecast

plt.plot(
    future_years,
    future_ensemble,
    marker="D",
    linestyle=":",
    linewidth=2,
    color="#8e44ad",
    label="Ensemble Forecast"
)

# Separate historical and future periods

plt.axvline(
    x=2024.5,
    color="gray",
    linestyle="--",
    linewidth=2,
    label="Forecast Boundary"
)

plt.axvspan(
    2024.5,
    2030.5,
    alpha=0.08,
    color="orange"
)

plt.xlabel("Year")

plt.ylabel("Mission Success Rate (%)")

plt.title(
    "Spacecraft Mission Success Rate "
    "Forecast (2025-2030)\n"
    "Experimental Projections - Not Validated"
)

plt.xticks(
    np.arange(2020, 2031)
)

plt.ylim(0, 110)

plt.legend()

plt.grid(
    alpha=0.3
)

save_and_show(
    "prediction_graph.png"
)

# ============================================================
# STEP 17: SAVE MODEL COMPARISON METRICS
# ============================================================

print("\n[15] Saving Model Comparison Metrics...")

metrics_df = pd.DataFrame({

    "Model": [
        "Random Forest",
        "Deep Learning"
    ],

    "Training_MAE": [
        rf_mae,
        dl_mae
    ],

    "Training_MSE": [
        rf_mse,
        dl_mse
    ],

    "Training_RMSE": [
        rf_rmse,
        dl_rmse
    ],

    "Training_R2": [
        rf_r2,
        dl_r2
    ]

})

metrics_path = os.path.join(
    PROJECT_PATH,
    "model_comparison_metrics.csv"
)

metrics_df.to_csv(
    metrics_path,
    index=False
)

print("\nModel Comparison Metrics:")
print(metrics_df)

# ============================================================
# STEP 18: FINAL PROJECT SUMMARY
# ============================================================

print("\n" + "=" * 70)

print("SPACECRAFT MISSION ANALYSIS COMPLETED")

print("=" * 70)

total_missions = df[
    "Total_Missions"
].sum()

successful_missions = df[
    "Successful_Missions"
].sum()

failed_missions = df[
    "Failed_Missions"
].sum()

overall_success_rate = (
    successful_missions / total_missions * 100
    if total_missions > 0
    else 0
)

print("\nDATASET SUMMARY")

print("Total Records:", len(df))

print("Total Missions:", total_missions)

print("Successful Missions:", successful_missions)

print("Failed Missions:", failed_missions)

print(
    "Overall Success Rate:",
    round(overall_success_rate, 2),
    "%"
)

print("\nMODEL TRAINING PERFORMANCE")

print(
    "Random Forest RMSE:",
    round(rf_rmse, 4)
)

print(
    "Deep Learning RMSE:",
    round(dl_rmse, 4)
)

print("\nGENERATED FILES")

generated_files = [
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "model_comparison_metrics.csv"
]

for filename in generated_files:

    filepath = os.path.join(
        PROJECT_PATH,
        filename
    )

    if os.path.exists(filepath):

        print("[SUCCESS]", filename)

    else:

        print("[FAILED]", filename)

print("\nAll files saved in:")

print(PROJECT_PATH)

print("\nIMPORTANT LIMITATIONS:")

print(
    "1. Model metrics describe training fit, "
    "not independent predictive accuracy."
)

print(
    "2. Forecasts are exploratory because "
    "only five annual records are available."
)

print(
    "3. Random Forest forecasts beyond "
    "the training period may remain constant."
)

print(
    "4. Neural network extrapolation "
    "may be unstable."
)

print(
    "5. The 2024 record represents "
    "a partial reporting period."
)

print("\n" + "=" * 70)
print("ALL GRAPHS AND CSV FILES GENERATED")
print("=" * 70)
```
