```python
import os
import json
import warnings
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import yaml

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ==============================================================
# STEP 1: PROJECT CONFIGURATION
# ==============================================================

PROJECT_PATH = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Spacecraft Mission Success Analysis and Prediction"
)

DATASET_PATH = os.path.join(
    PROJECT_PATH,
    "RS_Session_266_AU_2850_2.csv"
)

os.makedirs(PROJECT_PATH, exist_ok=True)

RANDOM_STATE = 42
rng = np.random.default_rng(RANDOM_STATE)

# AIS parameters
POPULATION_SIZE = 60
GENERATIONS = 250
SELECTED_ANTIBODIES = 12
CLONES_PER_ANTIBODY = 8
MUTATION_RATE = 0.15
RANDOM_REPLACEMENTS = 8

# Degree 2 polynomial regression
POLYNOMIAL_DEGREE = 2

# Penalize unnecessarily large coefficients
REGULARIZATION = 0.01

# Forecast horizon
FORECAST_START = 2025
FORECAST_END = 2030

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (12, 7)
plt.rcParams["font.size"] = 11

print("=" * 70)
print("ARTIFICIAL IMMUNE SYSTEM - SPACECRAFT ANALYSIS")
print("=" * 70)


# ==============================================================
# STEP 2: SAVE AND DISPLAY GRAPHS
# ==============================================================

def save_and_show(filename):

    filepath = os.path.join(PROJECT_PATH, filename)

    plt.tight_layout()

    plt.savefig(
        filepath,
        dpi=300,
        bbox_inches="tight"
    )

    print("[SAVED]", filepath)

    plt.show()
    plt.close()


# ==============================================================
# STEP 3: LOAD DATASET
# ==============================================================

print("\n[1] Loading Dataset...")

df = pd.read_csv(DATASET_PATH)

df.columns = df.columns.str.strip()

print("\nOriginal Columns:")
print(df.columns.tolist())


# ==============================================================
# STEP 4: DATA PREPROCESSING
# ==============================================================

print("\n[2] Preprocessing Dataset...")

df.rename(
    columns={
        "Total No. of Spacecraft Missions": "Total_Missions",
        "Successful Spacecraft Missions": "Successful_Missions",
        "Failed Spacecraft Missions": "Failed_Missions"
    },
    inplace=True
)

required_columns = [
    "Year",
    "Total_Missions",
    "Successful_Missions",
    "Failed_Missions"
]

missing = [
    col for col in required_columns
    if col not in df.columns
]

if missing:
    raise ValueError(
        f"Missing dataset columns: {missing}\n"
        f"Available columns: {df.columns.tolist()}"
    )

# Extract year from values such as 2024 (as on 10-12-24)
df["Year"] = (
    df["Year"]
    .astype(str)
    .str.extract(r"(\d{4})")[0]
)

for col in required_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df.dropna(subset=required_columns, inplace=True)
df.drop_duplicates(inplace=True)
df.sort_values("Year", inplace=True)
df.reset_index(drop=True, inplace=True)

df["Year"] = df["Year"].astype(int)

# Calculate historical success and failure rates
df["Success_Rate"] = np.where(
    df["Total_Missions"] > 0,
    df["Successful_Missions"] / df["Total_Missions"] * 100,
    0.0
)

df["Failure_Rate"] = np.where(
    df["Total_Missions"] > 0,
    df["Failed_Missions"] / df["Total_Missions"] * 100,
    0.0
)

if len(df) < 3:
    raise ValueError(
        "At least three annual observations are required."
    )

if df["Year"].duplicated().any():
    raise ValueError(
        "Multiple records exist for the same year. "
        "Aggregate them before running this analysis."
    )

print("\nProcessed Dataset:")
print(df.to_string(index=False))


# ==============================================================
# STEP 5: FEATURE ENGINEERING
# ==============================================================

print("\n[3] Preparing AIS Features...")

# Center and scale years to improve numerical stability.
# Example: 2020 -> -0.2, 2021 -> -0.1, etc.

YEAR_CENTER = float(df["Year"].mean())
YEAR_SCALE = 10.0

def create_features(years):

    years = np.asarray(years, dtype=float)

    normalized = (years - YEAR_CENTER) / YEAR_SCALE

    features = [
        np.ones_like(normalized)
    ]

    for degree in range(1, POLYNOMIAL_DEGREE + 1):
        features.append(normalized ** degree)

    return np.column_stack(features)


X = create_features(df["Year"].to_numpy())
y = df["Success_Rate"].to_numpy(dtype=float)

NUM_PARAMETERS = X.shape[1]

print("Feature Matrix Shape:", X.shape)
print("Target Shape:", y.shape)


# ==============================================================
# STEP 6: AIS FITNESS FUNCTION
# ==============================================================

print("\n[4] Defining AIS Fitness Function...")

def predict_raw(features, antibody):

    return features @ antibody


def predict_success_rate(features, antibody):

    predictions = predict_raw(features, antibody)

    return np.clip(predictions, 0, 100)


def calculate_fitness(antibody):

    predictions = predict_raw(X, antibody)

    mse = np.mean((y - predictions) ** 2)

    # Penalize polynomial coefficients, excluding intercept
    penalty = REGULARIZATION * np.sum(
        antibody[1:] ** 2
    )

    return float(mse + penalty)


# ==============================================================
# STEP 7: INITIALIZE ANTIBODY POPULATION
# ==============================================================

print("\n[5] Initializing AIS Population...")

# An antibody represents polynomial coefficients:
# [intercept, linear coefficient, quadratic coefficient]

population = rng.uniform(
    low=-50,
    high=50,
    size=(POPULATION_SIZE, NUM_PARAMETERS)
)

# Initialize one antibody with a constant mean baseline
population[0] = np.zeros(NUM_PARAMETERS)
population[0, 0] = float(np.mean(y))

# Initialize another using least squares
least_squares_solution = np.linalg.lstsq(
    X, y, rcond=None
)[0]

population[1] = least_squares_solution

best_antibody = None
best_fitness = float("inf")

fitness_history = []
mean_fitness_history = []


# ==============================================================
# STEP 8: CLONAL SELECTION ALGORITHM
# ==============================================================

print("\n[6] Running AIS Clonal Selection Optimization...")

for generation in range(GENERATIONS):

    # Evaluate antibody fitness
    fitness_values = np.array([
        calculate_fitness(antibody)
        for antibody in population
    ])

    # Rank population: lower fitness is better
    sorted_indices = np.argsort(fitness_values)

    population = population[sorted_indices]
    fitness_values = fitness_values[sorted_indices]

    # Store global best solution
    if fitness_values[0] < best_fitness:

        best_fitness = float(fitness_values[0])

        best_antibody = population[0].copy()

    fitness_history.append(best_fitness)
    mean_fitness_history.append(float(np.mean(fitness_values)))

    # Select the best antibodies
    selected = population[:SELECTED_ANTIBODIES]

    clones = []

    # Cloning and affinity-dependent hypermutation
    for rank, antibody in enumerate(selected):

        for clone_index in range(CLONES_PER_ANTIBODY):

            clone = antibody.copy()

            # Better-ranked antibodies receive smaller mutations.
            rank_factor = (
                (rank + 1) / SELECTED_ANTIBODIES
            )

            # Gradually reduce mutation magnitude
            progress_factor = (
                1.0 - generation / GENERATIONS
            )

            mutation_strength = (
                MUTATION_RATE
                * rank_factor
                * (0.1 + progress_factor)
                * 20
            )

            mutation = rng.normal(
                loc=0,
                scale=mutation_strength,
                size=NUM_PARAMETERS
            )

            clone += mutation

            clones.append(clone)

    clones = np.asarray(clones)

    # Random antibody replacements maintain diversity
    random_antibodies = rng.uniform(
        low=-50,
        high=50,
        size=(RANDOM_REPLACEMENTS, NUM_PARAMETERS)
    )

    # Combine old population, clones, random antibodies
    # and elite antibody
    candidates = np.vstack([
        population,
        clones,
        random_antibodies,
        best_antibody.reshape(1, -1)
    ])

    candidate_fitness = np.array([
        calculate_fitness(antibody)
        for antibody in candidates
    ])

    best_indices = np.argsort(
        candidate_fitness
    )[:POPULATION_SIZE]

    population = candidates[best_indices].copy()

    if (generation + 1) % 25 == 0:

        print(
            f"Generation {generation + 1:3d} | "
            f"Best Fitness: {best_fitness:.6f}"
        )

# Evaluate final population so the last generation is included
final_fitness = np.array([
    calculate_fitness(a) for a in population
])

final_index = int(np.argmin(final_fitness))

if final_fitness[final_index] < best_fitness:
    best_fitness = float(final_fitness[final_index])
    best_antibody = population[final_index].copy()

print("\nAIS Optimization Completed!")
print("Best Fitness:", best_fitness)
print("Best Antibody:", best_antibody)


# ==============================================================
# STEP 9: HISTORICAL FITTED VALUES
# ==============================================================

print("\n[7] Generating AIS Historical Predictions...")

ais_predictions = predict_success_rate(
    X,
    best_antibody
)

absolute_errors = np.abs(
    y - ais_predictions
)

squared_errors = (
    y - ais_predictions
) ** 2

ais_mae = mean_absolute_error(
    y, ais_predictions
)

ais_mse = mean_squared_error(
    y, ais_predictions
)

ais_rmse = np.sqrt(ais_mse)

ais_r2 = r2_score(
    y, ais_predictions
)

print("\nAIS Training Metrics:")
print("MAE:", ais_mae)
print("MSE:", ais_mse)
print("RMSE:", ais_rmse)
print("R2 Score:", ais_r2)

# Important:
# These are training-set metrics, NOT independent
# validation scores or classification accuracy.


# ==============================================================
# STEP 10: SAVE AIS MODEL
# ==============================================================

print("\n[8] Saving AIS Model...")

model_data = {
    "algorithm": "AIS_CLONALG",
    "model_type": "Polynomial Regression",
    "coefficients": best_antibody,
    "polynomial_degree": POLYNOMIAL_DEGREE,
    "year_center": YEAR_CENTER,
    "year_scale": YEAR_SCALE,
    "fitness": best_fitness,
    "training_years": df["Year"].tolist(),
    "training_only": True
}

model_path = os.path.join(
    PROJECT_PATH,
    "ais_model.pkl"
)

joblib.dump(model_data, model_path)

print("[SAVED]", model_path)


# ==============================================================
# STEP 11: AIS ACCURACY GRAPH
# ==============================================================

print("\n[9] Generating AIS Accuracy Graph...")

fig, axes = plt.subplots(
    1, 2,
    figsize=(15, 6)
)

# Optimization convergence
axes[0].plot(
    range(1, GENERATIONS + 1),
    fitness_history,
    color="#2980b9",
    linewidth=2,
    label="Best Fitness"
)

axes[0].plot(
    range(1, GENERATIONS + 1),
    mean_fitness_history,
    color="#e67e22",
    linewidth=1.5,
    alpha=0.7,
    label="Mean Population Fitness"
)

axes[0].set_title("AIS Optimization Convergence")
axes[0].set_xlabel("Generation")
axes[0].set_ylabel("Fitness (Regularized MSE)")
axes[0].legend()
axes[0].grid(alpha=0.3)

# Training errors
metric_names = ["MAE", "RMSE"]

metric_values = [ais_mae, ais_rmse]

bars = axes[1].bar(
    metric_names,
    metric_values,
    color=["#27ae60", "#e74c3c"],
    width=0.5
)

axes[1].set_title("AIS Training Error Metrics")
axes[1].set_ylabel("Error (Percentage Points)")
axes[1].set_ylim(
    0,
    max(metric_values + [1]) * 1.25
)

for bar in bars:

    height = bar.get_height()

    axes[1].text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"{height:.2f}",
        ha="center",
        va="bottom"
    )

fig.suptitle(
    "AIS Model Training Performance",
    fontsize=16
)

save_and_show("ais_accuracy_graph.png")


# ==============================================================
# STEP 12: AIS CORRELATION HEATMAP
# ==============================================================

print("\n[10] Generating AIS Heatmap...")

heatmap_columns = [
    "Year",
    "Total_Missions",
    "Successful_Missions",
    "Failed_Missions",
    "Success_Rate",
    "Failure_Rate"
]

correlation_matrix = df[
    heatmap_columns
].corr()

plt.figure(figsize=(11, 9))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    linewidths=0.5,
    square=True
)

plt.title(
    "AIS Spacecraft Mission Correlation Heatmap"
)

save_and_show("ais_heatmap.png")


# ==============================================================
# STEP 13: AIS COMPARISON GRAPH
# ==============================================================

print("\n[11] Generating AIS Comparison Graph...")

years = df["Year"].to_numpy()

plt.figure(figsize=(13, 7))

plt.plot(
    years,
    y,
    marker="o",
    color="black",
    linewidth=3,
    label="Actual Success Rate"
)

plt.plot(
    years,
    ais_predictions,
    marker="s",
    linestyle="--",
    color="#8e44ad",
    linewidth=2,
    label="AIS Fitted Success Rate"
)

plt.xlabel("Year")
plt.ylabel("Mission Success Rate (%)")

plt.title(
    "Actual vs AIS-Fitted Spacecraft "
    "Mission Success Rate"
)

plt.ylim(0, 110)
plt.xticks(years)
plt.legend()
plt.grid(alpha=0.3)

save_and_show("ais_comparison_graph.png")


# ==============================================================
# STEP 14: SAVE AIS RESULT CSV
# ==============================================================

print("\n[12] Generating AIS Result CSV...")

result_df = pd.DataFrame({

    "Year": df["Year"],

    "Total_Missions": df["Total_Missions"],

    "Successful_Missions": df["Successful_Missions"],

    "Failed_Missions": df["Failed_Missions"],

    "Actual_Success_Rate": df["Success_Rate"],

    "Actual_Failure_Rate": df["Failure_Rate"],

    "AIS_Predicted_Success_Rate": ais_predictions,

    "AIS_Absolute_Error": absolute_errors,

    "AIS_Squared_Error": squared_errors

})

result_df = result_df.round(4)

result_path = os.path.join(
    PROJECT_PATH,
    "ais_result.csv"
)

result_df.to_csv(
    result_path,
    index=False
)

print("[SAVED]", result_path)
print(result_df.to_string(index=False))


# ==============================================================
# STEP 15: AIS RESULT GRAPH
# ==============================================================

print("\n[13] Generating AIS Result Graph...")

successful = df[
    "Successful_Missions"
].to_numpy()

failed = df[
    "Failed_Missions"
].to_numpy()

plt.figure(figsize=(13, 7))

plt.bar(
    years,
    successful,
    color="#27ae60",
    label="Successful Missions"
)

plt.bar(
    years,
    failed,
    bottom=successful,
    color="#e74c3c",
    label="Failed Missions"
)

for i in range(len(years)):

    total = successful[i] + failed[i]

    plt.text(
        years[i],
        total + 0.1,
        str(int(total)),
        ha="center",
        fontweight="bold"
    )

plt.xlabel("Year")
plt.ylabel("Number of Missions")

plt.title(
    "AIS Project - Historical Spacecraft "
    "Mission Results"
)

plt.xticks(years)
plt.legend()
plt.grid(axis="y", alpha=0.3)

save_and_show("ais_result_graph.png")


# ==============================================================
# STEP 16: FUTURE PREDICTION USING AIS MODEL
# ==============================================================

print("\n[14] Generating AIS Future Predictions...")

future_years = np.arange(
    FORECAST_START,
    FORECAST_END + 1
)

future_X = create_features(
    future_years
)

future_predictions = predict_success_rate(
    future_X,
    best_antibody
)

print("\nExperimental Future Predictions:")

for year, prediction in zip(
    future_years,
    future_predictions
):

    print(
        f"{year}: {prediction:.2f}%"
    )


# ==============================================================
# STEP 17: SAVE AIS PREDICTION CSV
# ==============================================================

print("\n[15] Generating AIS Prediction CSV...")

prediction_df = pd.DataFrame({

    "Year": future_years,

    "AIS_Predicted_Success_Rate":
        np.round(future_predictions, 4),

    "Prediction_Type":
        ["Experimental Forecast"] * len(future_years),

    "Validated":
        [False] * len(future_years)

})

prediction_path = os.path.join(
    PROJECT_PATH,
    "ais_prediction.csv"
)

prediction_df.to_csv(
    prediction_path,
    index=False
)

print("[SAVED]", prediction_path)
print(prediction_df.to_string(index=False))


# ==============================================================
# STEP 18: AIS PREDICTION GRAPH
# ==============================================================

print("\n[16] Generating AIS Prediction Graph...")

plt.figure(figsize=(14, 8))

# Historical observations
plt.plot(
    years,
    y,
    marker="o",
    linewidth=3,
    color="black",
    label="Historical Success Rate"
)

# AIS future predictions
plt.plot(
    future_years,
    future_predictions,
    marker="s",
    linestyle="--",
    linewidth=2,
    color="#8e44ad",
    label="AIS Experimental Forecast"
)

# Forecast boundary
plt.axvline(
    x=2024.5,
    linestyle="--",
    color="gray",
    linewidth=2,
    label="Forecast Boundary"
)

plt.axvspan(
    2024.5,
    FORECAST_END + 0.5,
    color="orange",
    alpha=0.08
)

plt.xlabel("Year")
plt.ylabel("Mission Success Rate (%)")

plt.title(
    "AIS-Based Spacecraft Mission "
    "Success Rate Forecast (2025-2030)\n"
    "Experimental Projections - Not Validated"
)

plt.xticks(
    np.arange(
        int(years.min()),
        FORECAST_END + 1
    )
)

plt.ylim(0, 110)
plt.legend()
plt.grid(alpha=0.3)

save_and_show("ais_prediction_graph.png")


# ==============================================================
# STEP 19: SAVE AIS MODEL COMPARISON METRICS
# ==============================================================

print("\n[17] Saving AIS Evaluation Metrics...")

metrics_df = pd.DataFrame({

    "Model": ["AIS_CLONALG"],

    "Training_MAE": [ais_mae],

    "Training_MSE": [ais_mse],

    "Training_RMSE": [ais_rmse],

    "Training_R2": [ais_r2],

    "Best_Fitness": [best_fitness],

    "Generations": [GENERATIONS],

    "Population_Size": [POPULATION_SIZE],

    "Evaluation_Type": ["Training Only"]

})

metrics_path = os.path.join(
    PROJECT_PATH,
    "ais_model_comparison_metrics.csv"
)

metrics_df.to_csv(
    metrics_path,
    index=False
)

print("[SAVED]", metrics_path)


# ==============================================================
# STEP 20: SAVE AIS YAML CONFIGURATION
# ==============================================================

print("\n[18] Saving AIS Configuration...")

config = {

    "project": {
        "name": (
            "AI-Based Spacecraft Mission "
            "Success Analysis and Prediction"
        ),
        "algorithm": "Artificial Immune System",
        "optimization": "Clonal Selection Algorithm",
        "version": "1.0"
    },

    "dataset": {
        "filename": os.path.basename(DATASET_PATH),
        "records": int(len(df)),
        "features": ["Year"],
        "target": "Success_Rate"
    },

    "ais_parameters": {
        "population_size": POPULATION_SIZE,
        "generations": GENERATIONS,
        "selected_antibodies": SELECTED_ANTIBODIES,
        "clones_per_antibody": CLONES_PER_ANTIBODY,
        "mutation_rate": MUTATION_RATE,
        "random_replacements": RANDOM_REPLACEMENTS,
        "polynomial_degree": POLYNOMIAL_DEGREE,
        "regularization": REGULARIZATION,
        "random_state": RANDOM_STATE
    },

    "forecast": {
        "start_year": FORECAST_START,
        "end_year": FORECAST_END,
        "validated": False
    },

    "model": {
        "filename": "ais_model.pkl",
        "type": "AIS-Optimized Polynomial Regression"
    },

    "evaluation": {
        "type": "Training Only",
        "independent_validation": False
    }
}

yaml_path = os.path.join(
    PROJECT_PATH,
    "ais_config.yaml"
)

with open(
    yaml_path,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

print("[SAVED]", yaml_path)


# ==============================================================
# STEP 21: SAVE AIS JSON METADATA
# ==============================================================

print("\n[19] Saving AIS Metadata...")

total_missions = int(
    df["Total_Missions"].sum()
)

successful_missions = int(
    df["Successful_Missions"].sum()
)

failed_missions = int(
    df["Failed_Missions"].sum()
)

overall_success_rate = (
    successful_missions / total_missions * 100
    if total_missions > 0
    else 0.0
)

metadata = {

    "project_name": (
        "AI-Based Spacecraft Mission "
        "Success Analysis and Prediction"
    ),

    "algorithm": "AIS_CLONALG",

    "created_at": datetime.now().isoformat(),

    "dataset": {
        "records": int(len(df)),
        "start_year": int(df["Year"].min()),
        "end_year": int(df["Year"].max()),
        "total_missions": total_missions,
        "successful_missions": successful_missions,
        "failed_missions": failed_missions,
        "overall_success_rate": float(overall_success_rate)
    },

    "model": {
        "type": "Polynomial Regression",
        "optimization": "Clonal Selection",
        "coefficients": best_antibody.tolist(),
        "degree": POLYNOMIAL_DEGREE,
        "best_fitness": float(best_fitness)
    },

    "training_metrics": {
        "MAE": float(ais_mae),
        "MSE": float(ais_mse),
        "RMSE": float(ais_rmse),
        "R2": float(ais_r2)
    },

    "limitations": {
        "training_only": True,
        "validated_forecasts": False,
        "small_dataset": True,
        "partial_2024_observation": True,
        "note": (
            "Five annual observations are insufficient "
            "for reliable predictive validation. "
            "Forecasts are exploratory."
        )
    },

    "generated_files": [
        "ais_model.pkl",
        "ais_config.yaml",
        "ais_metadata.json",
        "ais_accuracy_graph.png",
        "ais_heatmap.png",
        "ais_comparison_graph.png",
        "ais_result.csv",
        "ais_result_graph.png",
        "ais_prediction.csv",
        "ais_prediction_graph.png",
        "ais_model_comparison_metrics.csv"
    ]
}

metadata_path = os.path.join(
    PROJECT_PATH,
    "ais_metadata.json"
)

with open(
    metadata_path,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4
    )

print("[SAVED]", metadata_path)


# ==============================================================
# STEP 22: FINAL VERIFICATION
# ==============================================================

print("\n" + "=" * 70)
print("AIS SPACECRAFT ANALYSIS COMPLETED")
print("=" * 70)

generated_files = [
    "ais_model.pkl",
    "ais_config.yaml",
    "ais_metadata.json",
    "ais_accuracy_graph.png",
    "ais_heatmap.png",
    "ais_comparison_graph.png",
    "ais_result.csv",
    "ais_result_graph.png",
    "ais_prediction.csv",
    "ais_prediction_graph.png",
    "ais_model_comparison_metrics.csv"
]

for filename in generated_files:

    filepath = os.path.join(
        PROJECT_PATH,
        filename
    )

    if os.path.exists(filepath):
        print("[SUCCESS]", filename)
    else:
        print("[FAILED]", filename)

print("\nDATASET SUMMARY")
print("Total Records:", len(df))
print("Total Missions:", total_missions)
print("Successful Missions:", successful_missions)
print("Failed Missions:", failed_missions)
print("Overall Success Rate:", round(overall_success_rate, 2), "%")

print("\nAIS TRAINING PERFORMANCE")
print("MAE:", round(ais_mae, 4))
print("MSE:", round(ais_mse, 4))
print("RMSE:", round(ais_rmse, 4))
print("R2:", round(ais_r2, 4))
print("Best Fitness:", round(best_fitness, 4))

print("\nAll files saved in:")
print(PROJECT_PATH)

print("\nIMPORTANT:")
print("Metrics represent training fit only.")
print("Future forecasts are exploratory and not validated.")

print("\n" + "=" * 70)
print("ALL AIS RESULTS GENERATED")
print("=" * 70)
```
